```python
from sklearn.datasets import load_breast_cancer
dataset = load_breast_cancer() # data: 569 samples, 30 features

dataset.feature_names, dataset.target_names
```

```python
x = dataset.data

stds = x.std(axis=0)
std_ratio = stds.max() / stds.min()
float(std_ratio) # we can see the standard deviation ratio (1 / 215170) is significant, so we need to normalize x
```

```python
means = x.mean(axis=0)
stds = x.std(axis=0)
x_normalized = (x - means) / stds # with gradient decent, normalizing makes sure that (with an std of 1 / 215170) big features do not have an unproportionate effect on the weighted sum

x_train = [list(map(float, sample)) for sample in x_normalized] # the dataset uses numpy array, our MLP uses list

x_train[:1]
```

```python
y = dataset.target

y_train = [1.0 if label == 1 else -1.0 for label in y] # target labels are 0/1, our model (using tanh) gives us -1/1

y_train[:20]
```
